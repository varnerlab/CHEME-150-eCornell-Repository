# K-Nearest Neighbors (KNN) Classification
In this module, we'll explore [K-nearest neighbor classification](https://en.wikipedia.org/wiki/K-nearest_neighbors_algorithm) and its kernelized variant, which are **supervised** approaches to solve binary and multi-class classification problems. By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* **K-nearest neighbor classification**: A supervised, non-parametric method (Fix & Hodges, 1951; Cover & Hart, 1967) that assigns a new point to the most common class among its $K$ nearest neighbors in the feature space. While KNN requires a labeled training set like other supervised methods, it differs by making no parametric assumptions about the data distribution and performs no explicit model training. Instead, it stores all training data and makes predictions based on local similarity using distance metrics such as [Euclidean distance](https://en.wikipedia.org/wiki/Euclidean_distance).

* **Kernelized K-nearest neighbor classification**: A variation of the standard [KNN algorithm](https://www.jstor.org/stable/1403797?socuuid=e7bcd649-778e-473f-8b7b-92805b5fab5f) that replaces traditional distance metrics with kernel similarity measures $k(\mathbf{x},\mathbf{x}^{\prime})$ to capture nonlinear relationships. This approach enables neighbor voting in a high-dimensional feature space without explicitly computing the transformed coordinates.

K-nearest neighbor classification is a flexible algorithm particularly useful for classification tasks where the decision boundary is complex or unknown. The kernelized version further enhances its capabilities by handling intricate nonlinear patterns that standard distance-based KNN might miss.

However, KNN illustrates a fundamental trade-off in machine learning: the **curse of dimensionality**. As the number of features increases, distance-based similarity becomes less meaningful because all points tend to appear nearly equidistant. This phenomenon can lead to degraded performance and requires careful consideration of feature selection and dimensionality reduction techniques.

## K-Nearest Neighbor Classification
K-nearest neighbor (KNN) classification is a simple yet powerful machine learning algorithm for classification and regression tasks. The algorithm finds the K closest data points to a new instance in the feature space and then classifies the new instance based on the majority class among these neighbors.

> __Key assumption__: The [K-nearest neighbor classifier](https://en.wikipedia.org/wiki/K-nearest_neighbors_algorithm) assumes that _similar_ inputs have _similar_ labels (in classification tasks) or _similar_ outputs for K-nearest neighbor regression tasks. However, the wiggly aspect of this assumption is what we mean by _similar_.

Let's sketch out the basic steps of KNN classification algorithm.

__Initialization__. Given a _reference_ dataset $\mathcal{D}_{R} = \{(\mathbf{x}_{i},y_{i}) \mid i = 1,2,\dots,n\}$, where $\mathbf{x}_i \in \mathbb{R}^{m}$ are $m$-dimensional feature vectors ($m\ll{n}$) and the target variables are discrete labels $y_i \in \left\{-1,1\right\}$. You will classify a new feature instance $\mathbf{z} \in \mathbb{R}^{m}$, where $\mathbf{z}\notin\mathcal{D}_{R}$, into one of the two classes $\{-1,1\}$. Initialize similarity map $\mathcal{S} \gets 0$.

1. For $i=1,2,\dots,n$ __do__:
    - Compute the distance (similarity) between the new feature vector $\mathbf{z}$ and each reference feature vector $\mathbf{x}_{i}$ in the reference dataset $\mathcal{D}_{R}$: $\mathcal{S}(i)\gets d(\mathbf{z},\mathbf{x}_{i})$, where $d(\cdot,\cdot)$ is a distance metric, e.g., Euclidean distance, Manhattan distance, etc.
    > __Kernelized variant__: Instead of using a distance metric, we can use a kernel function $k(\mathbf{z},\mathbf{x}_{i})$ to compute the similarity between the new feature vector $\mathbf{z}$ and each reference vector $\mathbf{x}_{i}$. The similarity map is then: $\mathcal{S}(i)\gets k(\mathbf{z},\mathbf{x}_{i})$.
2. Sort the map $\mathcal{S}$ from lowest to highest (ascending) distance. $S_{(1)} \ge S_{(2)} \ge \dots \ge S_{(n)}$, where ${(j)}$ denotes the index of the $j$-th smallest distance.
    > __Kernelized variant__: Sort the similarity map $\mathcal{S}$ from highest to lowest (descending) similarity. $S_{(1)} \le S_{(2)} \le \dots \le S_{(n)}$.
3. Select the top $K$ elements from the sorted similarity map $\mathcal{S}$, where $K$ is a hyperparameter that you can set based on your problem. Estimate the label of the new feature vector $\mathbf{z}$ by taking the majority vote among the labels of the top $K$ elements in the sorted similarity map $\mathcal{S}$: 
   - $\hat{y} = \text{argmax}_{y \in \{-1,1\}} \sum_{i=1}^{K} \mathbb{I}(y_i = y),$
   where $\mathbb{I}(\cdot)$ is the indicator function that returns 1 if the condition is true and 0 otherwise.

The space complexity of KNN is $\mathcal{O}(n)$ where $n$ is the size of the training dataset.

## What's good (or bad) about KNN?
Here are some **strengths** and **shortcomings** of the KNN algorithm we've proposed:

**Strengths**

* **Simplicity & Interpretability:** KNN is easy to understand and implement, with no complex training phase and minimal hyperparameters (primarily just the decision parameter $K$). The decision process is transparent—you can literally see which neighbors influenced each prediction.
* **Nonparametric Flexibility:** Because KNN makes no assumptions about the underlying data distribution, it can adapt to arbitrarily-shaped decision boundaries and capture complex patterns that parametric models (like linear classifiers) would miss.
* **Multi-class & Incremental Learning:** KNN generalizes to any number of classes without algorithmic changes. Adding new data points requires no retraining—simply insert them into the reference dataset.

**Shortcomings**

* **Computational Cost:** KNN is a _lazy learner_ that defers all computation until prediction time. Each query requires computing distances to all $n$ training points, making it $O(n)$ per prediction and prohibitively slow for large datasets.
* **Curse of Dimensionality:** In high-dimensional feature spaces, distances become less meaningful as neighbors tend to be nearly equidistant. This severely degrades performance without careful feature selection or dimensionality reduction.
* **Sensitivity Issues:** KNN is sensitive to: i) hyperparameter choice ($K$ and distance metric), ii) feature scaling (larger-range features dominate), iii) noisy outliers (especially with small $K$), and iv) class imbalance (bias toward majority classes in local neighborhoods).
