# K-Nearest Neighbors (KNN) Classification
In this module, we'll look at [K-nearest neighbor classification](https://en.wikipedia.org/wiki/K-nearest_neighbors_algorithm) and its kernelized variant, which is an _semi-supervised_ approach to solve (binary and beyond) classification problems. By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* __K-nearest neighbor classification__:  A _semi-supervised_ method (Fix & Hodges, 1951; Cover & Hart, 1967) that assigns a new point to the most common class among its $k$ nearest neighbors, typically using Euclidean distance. We say _semi-supervised_ because the algorithm requires a _labeled training set_ to determine the class of an (unlabeled) new point, but it does note require learning a model in the traditional sense (e.g., fitting parameters). Instead, it relies on the proximity of the new point to the labeled points in the training set to make its classification decision. Traditionally, distance in a KNN is measured as [the Euclidian distance](https://en.wikipedia.org/wiki/Euclidean_distance)
* __Kernelized K-nearest neighbor classification__: is a variation of the standard [KNN algorithm](https://www.jstor.org/stable/1403797?socuuid=e7bcd649-778e-473f-8b7b-92805b5fab5f) that replaces the Euclidean distance with a kernel similarity measure $k(\mathbf{x},\mathbf{x}^{\prime})$ to capture nonlinear structure. Thus, neighbors vote in a high-dimensional feature space without ever computing its coordinates.

K-nearest neighbor classification is a flexible algorithm useful for various classification tasks, especially when the data is not linearly separable. The kernelized version enhances its capabilities by allowing it to handle more complex relationships.

However, KNN illustrates a key trade-off in machine learning: the curse of dimensionality. As the number of features increases, the distance between points becomes less meaningful, making it harder to find meaningful neighbors. This can lead to overfitting and poor generalization performance.

## K-Nearest Neighbor Classification
K-nearest neighbor (KNN) classification is a simple yet powerful machine learning algorithm for classification and regression tasks. The algorithm finds the K closest data points to a new instance in the feature space and then classifies the new instance based on the majority class among these neighbors.

> __Key assumption__: The key assumption of a [K-nearest neighbor classifier](https://en.wikipedia.org/wiki/K-nearest_neighbors_algorithm) is that _similar_ inputs have _similar_ labels (in classification tasks) or _similar_ outputs for K-nearest neighbor regression tasks. However, the wiggly aspect of this assumption is what we mean by _similar_.

Let's sketch out the basic steps of KNN classification algorithm.

__Initialization__. Given a _reference_ dataset $\mathcal{D}_{R} = \{(\mathbf{x}_{i},y_{i}) \mid i = 1,2,\dots,n\}$, where $\mathbf{x}_i \in \mathbb{R}^{m}$ are $m$-dimensional feature vectors ($m\ll{n}$) and the target variables are discrete labels $y_i \in \left\{-1,1\right\}$. You will classify a new feature instance $\mathbf{z} \in \mathbb{R}^{m}$, where $\mathbf{z}\notin\mathcal{D}_{R}$, into one of the two classes $\{-1,1\}$. Initialize similarity map $\mathcal{S} \gets 0$.

1. For $i=1,2,\dots,n$ __do__:
    - Compute the distance (similarity) between the new feature vector $\mathbf{z}$ and each reference feature vector $\mathbf{x}_{i}$ in the reference dataset $\mathcal{D}_{R}$: $\mathcal{S}(i)\gets d(\mathbf{z},\mathbf{x}_{i})$.

2. Sort the similarity map $\mathcal{S}$ from highest to lowest similarity. $S_{(1)} \ge S_{(2)} \ge \dots \ge S_{(n)}$.
3. Select the top $K$ elements from the sorted similarity map $\mathcal{S}$, where $K$ is a hyperparameter that you can set based on your problem. Estimate the label of the new feature vector $\mathbf{z}$ by taking the majority vote among the labels of the top $K$ elements in the sorted similarity map $\mathcal{S}$: 
   - $$\hat{y} = \text{argmax}_{y \in \{-1,1\}} \sum_{i=1}^{K} \mathbb{I}(y_i = y),$$
   where $\mathbb{I}(\cdot)$ is the indicator function that returns 1 if the condition is true and 0 otherwise.
