# Linear Structured State Space Models of Sequences (S4)
This module introduces linear structured state space models of _long_ sequences. These models use a time-invariant linear state space representation of _hidden_ state dynamics, then some output mapping between the hidden state and the observed data. 

By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* __Linear time-invariant state space models__: Mathematical representations that describe how systems evolve over time using a hidden internal state. These models use fixed parameters to capture linear relationships between inputs, outputs, and internal states, making them powerful tools for sequence modeling.

* __S4 models__: Structured State Space sequence (S4) models combine classical state space methods with modern deep learning. S4 models use clever mathematical tricks, e.g., the [HiPPO framework](https://arxiv.org/abs/2206.12037), to efficiently process very long sequences that would be computationally expensive for traditional neural networks like transformers.

* __HiPPO framework__: HiPPO (High-order Polynomial Projection Operators) provides a smart way to remember long sequences by compressing past information into a fixed-size memory. Think of it as creating a "summary" of everything that happened before, where recent events get more detail while older events are compressed but not forgotten. This mathematical framework gives S4 models their ability to maintain useful information from very long sequences without overwhelming computational costs.

These models excel in tasks requiring handling extremely long sequences, such as those in the [Long Range Arena benchmark](https://arxiv.org/abs/2011.04006), by combining theoretically grounded memory mechanisms with computationally efficient implementations that can scale to long sequencess.
