# Linear Structured State Space Models of Sequences (S4)
This module introduces linear structured state space models of _long_ sequences. These models use a time-invariant linear state space representation of _hidden_ state dynamics, then some output mapping between the hidden state and the observed data. 

By the end of this module, you will be able to define and demonstrate mastery of the following key concepts:

* __Linear time-invariant state space models__: A time-invariant linear state-space model is a mathematical representation of a linear time-invariant (LTI) system using state variables, where the system's dynamics are described by four constant matrices $\mathbf{A}$, $\mathbf{B}$, $\mathbf{C}$, and $\mathbf{D}$. These models characterize systems with fixed parameters and linear relationships between inputs, outputs, and internal states over time.
* __S4 models__: Structured State Space sequence models (S4) are deep learning architectures that leverage the HiPPO (High-order Polynomial Projection Operators) framework for initialization, particularly HiPPO-LegS which uses Legendre polynomials with sliding window measures. S4 models address the computational bottleneck of naive state space implementations through structured parameterization of the state matrix $\mathbf{A}$, enabling efficient computation via diagonal plus low-rank decomposition and Fast Fourier Transform techniques.
* __HiPPO framework__: The HiPPO framework provides a principled approach to online compression of continuous signals and discrete time series through optimal polynomial projections. It produces optimal solutions for maintaining compressed representations of historical data, with HiPPO-LegS specifically designed to remember all history while avoiding assumptions about timescales.

These models excel in tasks requiring handling extremely long sequences, such as those in the [Long Range Arena benchmark](https://arxiv.org/abs/2011.04006), by combining theoretically grounded memory mechanisms with computationally efficient implementations that can scale to sequences of 10,000+ steps.
