# Derivation: The $\mathbf{U}$ Matrix and Conservation Relations in SVD of Stoichiometric Matrices

When we compute the SVD of a stoichiometric matrix $\mathbf{S}$, we obtain three factors: $\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top$. But which columns of $\mathbf{U}$ tell us about conservation relations—the invariant metabolite pools that remain constant in time? In this derivation, we show that the columns of $\mathbf{U}$ associated with zero singular values form an orthonormal basis for all conservation relations in the metabolic network.

___

## Geometry of the Stoichiometric Matrix and Subspaces

The stoichiometric matrix $\mathbf{S}\in\mathbb{R}^{m\times n}$ maps reaction flux vectors to metabolite net-production rates:
$$
\dot{\mathbf{x}} = \mathbf{S}\mathbf{v}
$$
where $\mathbf{x}\in\mathbb{R}^m$ is the vector of metabolite concentrations and $\mathbf{v}\in\mathbb{R}^n$ is the vector of reaction fluxes.

Any matrix partitions its codomain (the space of metabolites, $\mathbb{R}^m$) into two fundamental complementary subspaces:

**Column Space** $C(\mathbf{S})$: All vectors of the form $\mathbf{y} = \mathbf{S}\mathbf{v}$ for some $\mathbf{v}\in\mathbb{R}^n$. These are the achievable net production patterns for the metabolites.

**Left Nullspace** $N(\mathbf{S}^\top)$: All vectors $\mathbf{w}$ such that $\mathbf{S}^\top\mathbf{w} = \mathbf{0}$. These vectors have a special property: if $\mathbf{w}$ is in the left nullspace, then the weighted sum $\mathbf{w}^\top\mathbf{x}(t)$ never changes with time.

To see why, compute:
$$
\frac{d}{dt}(\mathbf{w}^\top\mathbf{x}) = \mathbf{w}^\top\dot{\mathbf{x}} = \mathbf{w}^\top(\mathbf{S}\mathbf{v}) = (\mathbf{w}^\top\mathbf{S})\mathbf{v} = (\mathbf{S}^\top\mathbf{w})^\top\mathbf{v} = \mathbf{0}^\top\mathbf{v} = 0
$$

The left nullspace of $\mathbf{S}$ is precisely where we find **conservation relations**—weighted sums of metabolites that are conserved in time. Biologically, these correspond to pools like adenylate (ATP+ADP+AMP), phosphate, and redox cofactors (NAD$^+$+NADH) that are preserved under the constraints of the stoichiometric matrix.

By the rank-nullity theorem, if $\text{rank}(\mathbf{S}) = r$, then:
$$
\dim(C(\mathbf{S})) + \dim(N(\mathbf{S}^\top)) = m
$$
$$
r + \dim(N(\mathbf{S}^\top)) = m
$$
$$
\dim(N(\mathbf{S}^\top)) = m - r
$$

So there are exactly $m-r$ linearly independent conservation relations.

___

## How SVD Reveals the Left Nullspace

The singular value decomposition factorizes $\mathbf{S}$ as:
$$
\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top
$$
where $\mathbf{U}\in\mathbb{R}^{m\times m}$ (or $\mathbf{U}\in\mathbb{R}^{m\times r}$ in thin SVD) is orthogonal with columns called **left singular vectors**, $\mathbf{\Sigma}\in\mathbb{R}^{m\times n}$ (or $\mathbf{\Sigma}\in\mathbb{R}^{r\times n}$) is diagonal with nonnegative singular values, and $\mathbf{V}$ is orthogonal.

In the full SVD, the $\mathbf{U}$ matrix has $m$ columns. We partition them based on whether the corresponding singular value is zero or nonzero:
$$
\mathbf{U} = \begin{bmatrix} \mathbf{U}_1 & \mathbf{U}_0 \end{bmatrix}
$$
where:
- $\mathbf{U}_1\in\mathbb{R}^{m\times r}$ contains columns associated with the $r$ nonzero singular values
- $\mathbf{U}_0\in\mathbb{R}^{m\times(m-r)}$ contains columns associated with the $m-r$ zero singular values

**Key Theorem**: The columns of $\mathbf{U}_0$ form an orthonormal basis for the left nullspace $N(\mathbf{S}^\top)$. That is:
$$
\boxed{\mathbf{S}^\top\mathbf{U}_0 = \mathbf{0}}
$$
Each column of $\mathbf{U}_0$ is itself a conservation relation, and any linear combination of these columns is also a conservation relation.

___

## Proof that $\mathbf{S}^\top\mathbf{U}_0 = \mathbf{0}$

We prove this by directly computing $\mathbf{S}^\top\mathbf{U}_0$ using the SVD factorization.

**Step 1: Transpose the SVD**

Starting from $\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top$, take the transpose:
$$
\mathbf{S}^\top = (\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top)^\top = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top
$$

**Step 2: Structure of $\mathbf{\Sigma}^\top$**

If $\mathbf{\Sigma}\in\mathbb{R}^{m\times n}$ is diagonal with singular values $\sigma_1, \sigma_2, \ldots, \sigma_r, 0, \ldots, 0$ on its diagonal (nonzero values first), then $\mathbf{\Sigma}^\top\in\mathbb{R}^{n\times m}$ has the block structure:
$$
\mathbf{\Sigma}^\top = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix}
$$
where $\mathbf{\Sigma}_r\in\mathbb{R}^{r\times r}$ is diagonal with the nonzero singular values.

**Step 3: Orthogonality property of $\mathbf{U}$**

Since the columns of $\mathbf{U}$ are orthonormal, we have $\mathbf{U}^\top\mathbf{U} = \mathbf{I}_m$. When we partition $\mathbf{U} = [\mathbf{U}_1 \; \mathbf{U}_0]$:
$$
\mathbf{U}^\top\mathbf{U}_0 = \begin{bmatrix} \mathbf{U}_1^\top \\ \mathbf{U}_0^\top \end{bmatrix}\mathbf{U}_0 = \begin{bmatrix} \mathbf{U}_1^\top\mathbf{U}_0 \\ \mathbf{U}_0^\top\mathbf{U}_0 \end{bmatrix} = \begin{bmatrix} \mathbf{0}_{r\times(m-r)} \\ \mathbf{I}_{m-r} \end{bmatrix}
$$
The first block is zero because columns in $\mathbf{U}_1$ and $\mathbf{U}_0$ are orthogonal to each other; the second block is identity because the columns of $\mathbf{U}_0$ are orthonormal.

**Step 4: Compute $\mathbf{S}^\top\mathbf{U}_0$**
$$
\mathbf{S}^\top\mathbf{U}_0 = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top\mathbf{U}_0 = \mathbf{V}\mathbf{\Sigma}^\top\begin{bmatrix} \mathbf{0}_{r\times(m-r)} \\ \mathbf{I}_{m-r} \end{bmatrix}
$$

Multiply $\mathbf{\Sigma}^\top$ by the partitioned vector:
$$
\mathbf{\Sigma}^\top\begin{bmatrix} \mathbf{0}_{r\times(m-r)} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix}\begin{bmatrix} \mathbf{0}_{r\times(m-r)} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{0}_{r\times(m-r)} \\ \mathbf{0}_{(m-r)\times(m-r)} \end{bmatrix} = \mathbf{0}
$$

The zero blocks in $\mathbf{\Sigma}^\top$ ensure that the entire product vanishes.

**Conclusion:**
$$
\mathbf{S}^\top\mathbf{U}_0 = \mathbf{V} \cdot \mathbf{0} = \mathbf{0} \quad\text{Q.E.D.}
$$

___

## Extracting Conservation Relations from SVD

In practice, we extract conservation relations from a stoichiometric matrix using this algorithm:

**Step 1:** Compute the SVD of the stoichiometric matrix:
$$
\mathbf{U}, \mathbf{\Sigma}, \mathbf{V} = \text{svd}(\mathbf{S})
$$

**Step 2:** Due to floating-point arithmetic, "zero" singular values are only approximately zero. Set a numerical tolerance:
$$
\tau = \epsilon \cdot \max(m, n) \cdot \sigma_1
$$
where $\epsilon\approx 10^{-15}$ is machine precision and $\sigma_1$ is the largest (first) singular value.

**Step 3:** Identify which singular values are negligible. Find all indices $i$ where $\sigma_i \le \tau$. These correspond to the conservation relations.

**Step 4:** Extract the columns of $\mathbf{U}$ corresponding to these small singular values. Call this extracted set $\mathbf{U}_0$. The columns of $\mathbf{U}_0$ form an orthonormal basis for all conservation relations.

**Step 5:** Any linear combination:
$$
\mathbf{w} = \mathbf{U}_0\,\mathbf{a}
$$
where $\mathbf{a}\in\mathbb{R}^{m-r}$ is a nonzero vector, is a valid conservation relation. The fact that $\mathbf{S}^\top\mathbf{w} = \mathbf{S}^\top(\mathbf{U}_0\,\mathbf{a}) = (\mathbf{S}^\top\mathbf{U}_0)\,\mathbf{a} = \mathbf{0}\,\mathbf{a} = \mathbf{0}$ ensures each such combination satisfies the conservation property.

___

## Summary

The SVD automatically partitions metabolite space into two fundamental subspaces: the achievable production patterns (spanned by columns of $\mathbf{U}$ with nonzero singular values) and the conservation relations (spanned by columns of $\mathbf{U}$ with zero singular values). Because columns of $\mathbf{U}$ associated with zero singular values satisfy $\mathbf{S}^\top\mathbf{U}_0 = \mathbf{0}$, they are exactly the vectors defining conserved metabolite pools.

> __Key Results:__
>
> * **Conservation relations are left nullspace vectors:** Any weighted sum $\mathbf{w}^\top\mathbf{x}$ is conserved if and only if $\mathbf{S}^\top\mathbf{w} = \mathbf{0}$, meaning $\mathbf{w}$ is in the left nullspace of the stoichiometric matrix
> * **SVD reveals all conservations:** The columns of $\mathbf{U}_0$ (those with zero singular values) form an orthonormal basis spanning all conservation relations without manual specification
> * **Simple algorithmic extraction:** Identify singular values below numerical tolerance, extract corresponding columns of $\mathbf{U}$, and any linear combination yields a valid conservation relation

## Proof that $\mathbf{S}^\top \mathbf{U}_0 = \mathbf{0}$

**Step 1: Transpose the SVD**
$$\mathbf{S}^\top = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top$$

**Step 2: Partition $\mathbf{\Sigma}^\top$**
$$\mathbf{\Sigma}^\top = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix}$$
where $\mathbf{\Sigma}_r$ is $r \times r$ with nonzero singular values.

**Step 3: Compute $\mathbf{S}^\top \mathbf{U}_0$**
$$\mathbf{S}^\top \mathbf{U}_0 = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top \mathbf{U}_0$$

Since $\mathbf{U} = [\mathbf{U}_1 \; \mathbf{U}_0]$ is orthonormal:
$$\mathbf{U}^\top \mathbf{U}_0 = \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix}$$

**Step 4: Apply zero block**
$$\mathbf{\Sigma}^\top \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix} \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{0} \\ \mathbf{0} \end{bmatrix}$$

**Conclusion:**
$$\boxed{\mathbf{S}^\top \mathbf{U}_0 = \mathbf{V} \begin{bmatrix} \mathbf{0} \\ \mathbf{0} \end{bmatrix} = \mathbf{0}} \quad\text{Q.E.D.}$$

The zero block in $\mathbf{\Sigma}^\top$ ensures the product vanishes.

Given the SVD of $\mathbf{S}$:

SVD partitions the metabolite space into two subspaces: achievable production patterns (nonzero singular values) and conservation relations (zero singular values). The left singular vectors with zero singular values form an orthonormal basis for all conserved metabolite pools.

# Derivation: The $\mathbf{U}$ Matrix and Conservation Relations in SVD of Stoichiometric Matrices

In this notebook, we develop a mathematical understanding of the left singular vectors ($\mathbf{U}$ matrix) from the SVD of a stoichiometric matrix $\mathbf{S}$. We'll build from basic geometry through a complete derivation, showing how SVD reveals the **conservation relations** encoded in metabolic networks.

> __Learning Objectives:__
>
> * __Understand metabolite space partitions:__ Explain how the stoichiometric matrix partitions $\mathbb{R}^m$ into the column space and left nullspace
> * __Connect SVD structure to conservation relations:__ Show why columns of $\mathbf{U}$ with zero singular values form a basis for conservation relations
> * __Prove the nullspace property:__ Derive and verify that $\mathbf{S}^\top\mathbf{U}_0 = \mathbf{0}$

___

## Part 1: The Geometry of the Stoichiometric Matrix

Recall that the stoichiometric matrix $\mathbf{S} \in \mathbb{R}^{m \times n}$ maps reaction flux vectors to metabolite net-production vectors:
$$
\dot{\mathbf{x}} = \mathbf{S}\mathbf{v}
$$

where:
* $\mathbf{x} \in \mathbb{R}^m$ is the metabolite concentration vector ($m$ metabolites)
* $\mathbf{v} \in \mathbb{R}^n$ is the reaction flux vector ($n$ reactions)
* $\dot{\mathbf{x}}$ is the time derivative of metabolite concentrations

### Subspaces of Interest

Any matrix $\mathbf{S}$ partitions $\mathbb{R}^m$ into complementary subspaces:

| Subspace | Definition | Meaning |
|-----------|-----------|----------|
| **Column space** $C(\mathbf{S})$ | All vectors $\mathbf{y} = \mathbf{S}\mathbf{v}$ for some $\mathbf{v}$ | Achievable net metabolite patterns |
| **Left nullspace** $N(\mathbf{S}^\top)$ | All vectors $\mathbf{w}$ such that $\mathbf{S}^\top\mathbf{w} = \mathbf{0}$ | **Conservation relations** |
| **Row space** $C(\mathbf{S}^\top)$ | Orthogonal complement of left nullspace | Span of matrix rows |
| **Right nullspace** $N(\mathbf{S})$ | All vectors $\mathbf{v}$ such that $\mathbf{S}\mathbf{v} = \mathbf{0}$ | Steady-state flux patterns |

By the rank-nullity theorem:
- $\dim(C(\mathbf{S})) + \dim(N(\mathbf{S}^\top)) = m$
- If $\text{rank}(\mathbf{S}) = r$, then $\dim(C(\mathbf{S})) = r$ and $\dim(N(\mathbf{S}^\top)) = m - r$

___

## Part 2: Understanding Conservation Relations

### What is a Conservation Relation?

A **conservation relation** is a weighted sum of metabolites whose total amount is invariant in time.

Mathematically, if $\mathbf{w}^\top \mathbf{S} = \mathbf{0}^\top$, then:
$$
\frac{d}{dt}(\mathbf{w}^\top \mathbf{x}) = \mathbf{w}^\top \dot{\mathbf{x}} = \mathbf{w}^\top (\mathbf{S}\mathbf{v}) = (\mathbf{w}^\top\mathbf{S})\mathbf{v} = \mathbf{0}
$$

Therefore, $\mathbf{w}^\top \mathbf{x}(t)$ is constant for all $t$. The weighted sum of metabolites specified by $\mathbf{w}$ never changes.

### Biological Examples

* **Adenylate pool**: Total ATP + ADP + AMP is conserved (conversions between forms preserve total count)
* **Phosphate**: Total phosphate across nucleotides and phosphorylated metabolites
* **Redox cofactors**: NAD$^+$ + NADH and NADP$^+$ + NADPH (in redox-balance assumptions)

### Equivalence to Left Nullspace

The condition $\mathbf{w}^\top \mathbf{S} = \mathbf{0}^\top$ is equivalent to:
$$
\mathbf{S}^\top \mathbf{w} = \mathbf{0}
$$

Conservation relations are exactly the vectors in the left nullspace of $\mathbf{S}$: $N(\mathbf{S}^\top)$.

___

## Part 3: How SVD Reveals the Left Nullspace

### Partitioning the SVD

Recall the SVD decomposition:
$$
\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top
$$

The $\mathbf{U}$ matrix ($m \times m$ in full SVD, or $m \times r$ in thin SVD) has orthonormal columns that form a basis for $\mathbb{R}^m$. We partition $\mathbf{U}$ based on singular value magnitude:
$$
\mathbf{U} = \begin{bmatrix} \mathbf{U}_1 & \mathbf{U}_0 \end{bmatrix}
$$

where:
* $\mathbf{U}_1 \in \mathbb{R}^{m \times r}$ contains the first $r$ columns (associated with nonzero singular values)
* $\mathbf{U}_0 \in \mathbb{R}^{m \times (m-r)}$ contains the remaining $m-r$ columns (associated with zero singular values)

### Key Geometric Fact

**Theorem**: The columns of $\mathbf{U}_0$ form an orthonormal basis for the left nullspace $N(\mathbf{S}^\top)$.

In other words:
$$
\mathbf{S}^\top \mathbf{U}_0 = \mathbf{0}
$$

Each column of $\mathbf{U}_0$ is a conservation relation. Any linear combination of these columns is also a conservation relation.

___

## Part 4: Mathematical Proof

We prove that $\mathbf{S}^\top \mathbf{U}_0 = \mathbf{0}$.

### Step 1: Express $\mathbf{S}^\top$ Using SVD

Starting from $\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top$, we compute:
$$
\mathbf{S}^\top = \left(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top\right)^\top = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top
$$

### Step 2: Characterize $\mathbf{\Sigma}^\top$

If $\mathbf{\Sigma}$ is $m \times n$ with singular values on its diagonal, then $\mathbf{\Sigma}^\top$ is $n \times m$. Partitioning to match $\mathbf{U}$ and $\mathbf{U}_0$:
$$
\mathbf{\Sigma}^\top = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix}
$$

where $\mathbf{\Sigma}_r$ is $r \times r$ with nonzero singular values on its diagonal.

### Step 3: Compute $\mathbf{S}^\top \mathbf{U}_0$

Multiply $\mathbf{S}^\top$ by $\mathbf{U}_0$:
$$
\mathbf{S}^\top \mathbf{U}_0 = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top \mathbf{U}_0
$$

Since $\mathbf{U} = \begin{bmatrix} \mathbf{U}_1 & \mathbf{U}_0 \end{bmatrix}$ is orthonormal:
$$
\mathbf{U}^\top \mathbf{U}_0 = \begin{bmatrix} \mathbf{U}_1^\top \\ \mathbf{U}_0^\top \end{bmatrix} \mathbf{U}_0 = \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix}
$$

### Step 4: Apply the Zero Block

Multiply by $\mathbf{\Sigma}^\top$:
$$
\mathbf{\Sigma}^\top \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix} \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{0} \\ \mathbf{0} \end{bmatrix}
$$

The zero block in $\mathbf{\Sigma}^\top$ annihilates the product.

### Step 5: Final Result

$$
\mathbf{S}^\top \mathbf{U}_0 = \mathbf{V} \begin{bmatrix} \mathbf{0} \\ \mathbf{0} \end{bmatrix} = \mathbf{0}
$$

**Q.E.D.** ✓

## Part 5: Practical Algorithm

Given the SVD of $\mathbf{S}$, we extract conservation relations as follows.

**Step 1: Compute SVD**
$$
\mathbf{U}, \mathbf{\Sigma}, \mathbf{V} = \text{svd}(\mathbf{S})

**Step 2: Set Numerical Tolerance**

$$
\mathbf{S}^\top \mathbf{w} = \mathbf{S}^\top (\mathbf{U}_0 \mathbf{a}) = (\mathbf{S}^\top \mathbf{U}_0) \mathbf{a} = \mathbf{0} \cdot \mathbf{a} = \mathbf{0}
$$
