# Derivation: The $\mathbf{U}$ Matrix and Conservation Relations in SVD of Stoichiometric Matrices

In this notebook, we develop a mathematical understanding of the left singular vectors ($\mathbf{U}$ matrix) from the SVD of a stoichiometric matrix $\mathbf{S}$. We'll build from basic geometry through a complete derivation, showing how SVD reveals the **conservation relations** encoded in metabolic networks.

> __Learning Objectives:__
>
> * __Understand metabolite space partitions:__ Explain how the stoichiometric matrix partitions $\mathbb{R}^m$ into the column space and left nullspace
> * __Connect SVD structure to conservation relations:__ Show why columns of $\mathbf{U}$ with zero singular values form a basis for conservation relations
> * __Prove the nullspace property:__ Derive and verify that $\mathbf{S}^\top\mathbf{U}_0 = \mathbf{0}$

___

## Part 1: The Geometry of the Stoichiometric Matrix

Recall that the stoichiometric matrix $\mathbf{S} \in \mathbb{R}^{m \times n}$ maps reaction flux vectors to metabolite net-production vectors:
$$
\dot{\mathbf{x}} = \mathbf{S}\mathbf{v}
$$

where:
* $\mathbf{x} \in \mathbb{R}^m$ is the metabolite concentration vector ($m$ metabolites)
* $\mathbf{v} \in \mathbb{R}^n$ is the reaction flux vector ($n$ reactions)
* $\dot{\mathbf{x}}$ is the time derivative of metabolite concentrations

### Subspaces of Interest

Any matrix $\mathbf{S}$ partitions $\mathbb{R}^m$ into complementary subspaces:

| Subspace | Definition | Meaning |
|-----------|-----------|----------|
| **Column space** $C(\mathbf{S})$ | All vectors $\mathbf{y} = \mathbf{S}\mathbf{v}$ for some $\mathbf{v}$ | Achievable net metabolite patterns |
| **Left nullspace** $N(\mathbf{S}^\top)$ | All vectors $\mathbf{w}$ such that $\mathbf{S}^\top\mathbf{w} = \mathbf{0}$ | **Conservation relations** |
| **Row space** $C(\mathbf{S}^\top)$ | Orthogonal complement of left nullspace | Span of matrix rows |
| **Right nullspace** $N(\mathbf{S})$ | All vectors $\mathbf{v}$ such that $\mathbf{S}\mathbf{v} = \mathbf{0}$ | Steady-state flux patterns |

By the rank-nullity theorem:
- $\dim(C(\mathbf{S})) + \dim(N(\mathbf{S}^\top)) = m$
- If $\text{rank}(\mathbf{S}) = r$, then $\dim(C(\mathbf{S})) = r$ and $\dim(N(\mathbf{S}^\top)) = m - r$

___

## Part 2: Understanding Conservation Relations

### What is a Conservation Relation?

A **conservation relation** is a weighted sum of metabolites whose total amount is invariant in time.

Mathematically, if $\mathbf{w}^\top \mathbf{S} = \mathbf{0}^\top$, then:
$$
\frac{d}{dt}(\mathbf{w}^\top \mathbf{x}) = \mathbf{w}^\top \dot{\mathbf{x}} = \mathbf{w}^\top (\mathbf{S}\mathbf{v}) = (\mathbf{w}^\top\mathbf{S})\mathbf{v} = \mathbf{0}
$$

Therefore, $\mathbf{w}^\top \mathbf{x}(t)$ is constant for all $t$. The weighted sum of metabolites specified by $\mathbf{w}$ never changes.

### Biological Examples

* **Adenylate pool**: Total ATP + ADP + AMP is conserved (conversions between forms preserve total count)
* **Phosphate**: Total phosphate across nucleotides and phosphorylated metabolites
* **Redox cofactors**: NAD$^+$ + NADH and NADP$^+$ + NADPH (in redox-balance assumptions)

### Equivalence to Left Nullspace

The condition $\mathbf{w}^\top \mathbf{S} = \mathbf{0}^\top$ is equivalent to:
$$
\mathbf{S}^\top \mathbf{w} = \mathbf{0}
$$

Conservation relations are exactly the vectors in the left nullspace of $\mathbf{S}$: $N(\mathbf{S}^\top)$.

___

## Part 3: How SVD Reveals the Left Nullspace

### Partitioning the SVD

Recall the SVD decomposition:
$$
\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top
$$

The $\mathbf{U}$ matrix ($m \times m$ in full SVD, or $m \times r$ in thin SVD) has orthonormal columns that form a basis for $\mathbb{R}^m$. We partition $\mathbf{U}$ based on singular value magnitude:
$$
\mathbf{U} = \begin{bmatrix} \mathbf{U}_1 & \mathbf{U}_0 \end{bmatrix}
$$

where:
* $\mathbf{U}_1 \in \mathbb{R}^{m \times r}$ contains the first $r$ columns (associated with nonzero singular values)
* $\mathbf{U}_0 \in \mathbb{R}^{m \times (m-r)}$ contains the remaining $m-r$ columns (associated with zero singular values)

### Key Geometric Fact

**Theorem**: The columns of $\mathbf{U}_0$ form an orthonormal basis for the left nullspace $N(\mathbf{S}^\top)$.

In other words:
$$
\mathbf{S}^\top \mathbf{U}_0 = \mathbf{0}
$$

Each column of $\mathbf{U}_0$ is a conservation relation. Any linear combination of these columns is also a conservation relation.

___

## Part 4: Mathematical Proof

We prove that $\mathbf{S}^\top \mathbf{U}_0 = \mathbf{0}$.

### Step 1: Express $\mathbf{S}^\top$ Using SVD

Starting from $\mathbf{S} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top$, we compute:
$$
\mathbf{S}^\top = \left(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top\right)^\top = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top
$$

### Step 2: Characterize $\mathbf{\Sigma}^\top$

If $\mathbf{\Sigma}$ is $m \times n$ with singular values on its diagonal, then $\mathbf{\Sigma}^\top$ is $n \times m$. Partitioning to match $\mathbf{U}$ and $\mathbf{U}_0$:
$$
\mathbf{\Sigma}^\top = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix}
$$

where $\mathbf{\Sigma}_r$ is $r \times r$ with nonzero singular values on its diagonal.

### Step 3: Compute $\mathbf{S}^\top \mathbf{U}_0$

Multiply $\mathbf{S}^\top$ by $\mathbf{U}_0$:
$$
\mathbf{S}^\top \mathbf{U}_0 = \mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top \mathbf{U}_0
$$

Since $\mathbf{U} = \begin{bmatrix} \mathbf{U}_1 & \mathbf{U}_0 \end{bmatrix}$ is orthonormal:
$$
\mathbf{U}^\top \mathbf{U}_0 = \begin{bmatrix} \mathbf{U}_1^\top \\ \mathbf{U}_0^\top \end{bmatrix} \mathbf{U}_0 = \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix}
$$

### Step 4: Apply the Zero Block

Multiply by $\mathbf{\Sigma}^\top$:
$$
\mathbf{\Sigma}^\top \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{\Sigma}_r & \mathbf{0} \\ \mathbf{0} & \mathbf{0} \end{bmatrix} \begin{bmatrix} \mathbf{0} \\ \mathbf{I}_{m-r} \end{bmatrix} = \begin{bmatrix} \mathbf{0} \\ \mathbf{0} \end{bmatrix}
$$

The zero block in $\mathbf{\Sigma}^\top$ annihilates the product.

### Step 5: Final Result

$$
\mathbf{S}^\top \mathbf{U}_0 = \mathbf{V} \begin{bmatrix} \mathbf{0} \\ \mathbf{0} \end{bmatrix} = \mathbf{0}
$$

**Q.E.D.** ✓

___

## Part 5: Practical Algorithm

Given the SVD of $\mathbf{S}$, we extract conservation relations as follows.

**Step 1: Compute SVD**
$$
\mathbf{U}, \mathbf{\Sigma}, \mathbf{V} = \text{svd}(\mathbf{S})
$$

**Step 2: Set Numerical Tolerance**
$$
\tau = \epsilon \cdot \max(m, n) \cdot \sigma_1
$$

where $\epsilon$ is machine precision ($\sim 10^{-15}$) and $\sigma_1$ is the largest singular value.

**Step 3: Identify Zero Singular Values**

Find all indices $i$ where $\sigma_i \leq \tau$.

**Step 4: Extract Basis Vectors**

The columns of $\mathbf{U}$ corresponding to these indices form $\mathbf{U}_0$, an orthonormal basis for all conservation relations.

**Step 5: Construct Conservation Relations**

Any linear combination:
$$
\mathbf{w} = \mathbf{U}_0 \mathbf{a}
$$

for nonzero coefficient vector $\mathbf{a}$ is a valid conservation relation. This is because:
$$
\mathbf{S}^\top \mathbf{w} = \mathbf{S}^\top (\mathbf{U}_0 \mathbf{a}) = (\mathbf{S}^\top \mathbf{U}_0) \mathbf{a} = \mathbf{0} \cdot \mathbf{a} = \mathbf{0}
$$

___

## Summary

The left singular vectors from SVD partition metabolite space into two complementary subspaces: achievable production patterns (nonzero singular values) and conservation relations (zero singular values). SVD automatically discovers all conserved metabolite pools without additional computation.

> __Key Takeaways:__
>
> * **Conservation relations are left nullspace vectors:** A weighted sum $\mathbf{w}^\top \mathbf{x}$ is conserved if and only if $\mathbf{S}^\top \mathbf{w} = \mathbf{0}$
> * **SVD partitions reveal conservation:** The columns of $\mathbf{U}$ associated with zero singular values form an orthonormal basis for all conservation relations
> * **Practical extraction is straightforward:** Identify singular values below numerical tolerance, extract corresponding columns of $\mathbf{U}$, and any linear combination yields a conservation relation
