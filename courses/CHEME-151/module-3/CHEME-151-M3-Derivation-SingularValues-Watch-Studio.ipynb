# Derivation: Relationship between the Singular Values and the Eigenvalues
One obvious question is how are the singular values related to the eigenvalues of a matrix? Assume $\mathbf{A}\in\mathbb{R}^{m\times n}$ has the singular value decomposition (SVD):
$$
\begin{align*}
\mathbf{A} &=\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top}\\
\end{align*}
$$
with $\mathbf{U}\in\mathbb{R}^{m\times m}$ orthogonal, $\mathbf{V}\in\mathbb{R}^{n\times n}$ orthogonal, and $\mathbf{\Sigma}\in\mathbb{R}^{m\times n}$ is diagonal (nonnegative entries on its main diagonal). Let $p=\min(m,n)$ and denote the singular values as $\sigma_1,\dots,\sigma_p$.

The singular values, $\sigma_i$, are related to the eigenvalues, $\lambda_i$, of the matrices $\mathbf{A}^\top \mathbf{A}$ and $\mathbf{A}\mathbf{A}^\top$. Further, the columns of $\mathbf{V}$ (right singular vectors) are the eigenvectors of $\mathbf{A}^\top \mathbf{A}$ and the columns of $\mathbf{U}$ (left singular vectors) are the eigenvectors of $\mathbf{A}\mathbf{A}^\top$. Where does this come from?

### Right singular vectors and eigenvalues of $\mathbf{A}^\top \mathbf{A}$
Start by forming the (square) $\mathbf{A}^\top\mathbf{A}\in\mathbb{R}^{n\times n}$ and substitute the SVD decomposition:
$$
\begin{align*}
\mathbf{A}^\top\mathbf{A} &= (\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top})^\top(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top})\\
&= \mathbf{V}\mathbf{\Sigma}^{\top}\underbrace{\left(\mathbf{U}^\top\mathbf{U}\right)}_{\mathbf{I}_{m}}\mathbf{\Sigma}\mathbf{V}^{\top}\\
&= \mathbf{V}\left(\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\right)\mathbf{V}^{\top}\\
\end{align*}
$$
Note that $\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\in\mathbb{R}^{n\times n}$ is a diagonal matrix with diagonal entries $\sigma_1^2,\dots,\sigma_p^2$ (the squares of the singular values) followed by $n-p$ zeros:
$$
\begin{align*}
\mathbf{\Sigma}^{\top}\mathbf{\Sigma} = \begin{bmatrix}
\sigma_1^2 & 0 & \cdots & 0\\
0 & \sigma_2^2 & \cdots & 0\\
\vdots & \vdots & \ddots & \vdots\\
0 & 0 & \cdots & 0
\end{bmatrix}
\end{align*}
$$
or in a more compact notation $\mathbf{\Sigma}^{\top}\mathbf{\Sigma} = \mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)$. Given this, we can see that:
$$
\begin{align*}
\mathbf{A}^\top\mathbf{A}\mathbf{V} &= \mathbf{V}\left(\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\right)\underbrace{\mathbf{V}^{\top}\mathbf{V}}_{\mathbf{I}_n}\\
& = \mathbf{V}\left(\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\right)\\
& = \mathbf{V\mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)}\\
& = \mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)\mathbf{V}\\
& = \mathbf{\Lambda}\mathbf{V}
\end{align*}
$$
where $\mathbf{\Lambda}=\mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)$ is the diagonal matrix of eigenvalues of $\mathbf{A}^\top \mathbf{A}$. This shows that the columns of $\mathbf{V}$ are the eigenvectors of $\mathbf{A}^\top \mathbf{A}$ and the corresponding eigenvalues are the squares of the singular values, $\lambda_i=\sigma_i^2$ for $i=1,\dots,p$, with the remaining $n-p$ eigenvalues being zero.




$$
\mathbf{A}^\top\mathbf{A}
=========================

# (\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top)^\top(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top)

(\mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top)(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top).
$$

Use orthogonality: $\mathbf{U}^\top\mathbf{U}=\mathbf{I}_m$:

$$
\mathbf{A}^\top\mathbf{A}
=========================

# \mathbf{V}\mathbf{\Sigma}^\top(\mathbf{U}^\top\mathbf{U})\mathbf{\Sigma}\mathbf{V}^\top

\mathbf{V}(\mathbf{\Sigma}^\top\mathbf{\Sigma})\mathbf{V}^\top.
$$

Now analyze $\mathbf{\Sigma}^\top\mathbf{\Sigma}$.

Since $\mathbf{\Sigma}$ is rectangular diagonal, it has the form (conceptually)
$$
\mathbf{\Sigma}=
\begin{bmatrix}
\operatorname{diag}(\sigma_1,\dots,\sigma_p)\
\mathbf{0}
\end{bmatrix}
\quad \text{or} \quad
\mathbf{\Sigma}=
\begin{bmatrix}
\operatorname{diag}(\sigma_1,\dots,\sigma_p) & \mathbf{0}
\end{bmatrix},
$$
depending on whether $m\ge n$ or $m<n$. In either case,
$$
\mathbf{\Sigma}^\top\mathbf{\Sigma}\in\mathbb{R}^{n\times n}
\quad\text{is diagonal with entries}\quad
\sigma_1^2,\dots,\sigma_p^2,
$$
and if $n>p$ the remaining diagonal entries are $0$.

So we can write
$$
\mathbf{\Sigma}^\top\mathbf{\Sigma}
===================================

\operatorname{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)\in\mathbb{R}^{n\times n}.
$$

Putting it together:
$$
\mathbf{A}^\top\mathbf{A}
=========================

\mathbf{V},\operatorname{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0),\mathbf{V}^\top.
$$

**Conclusion (right side):**

* The eigenvalues of $\mathbf{A}^\top\mathbf{A}$ are $\sigma_1^2,\dots,\sigma_p^2$ plus zeros (if needed to make $n$ eigenvalues total).
* The eigenvectors are the columns of $\mathbf{V}$.
* In particular, for each $i\le p$,
  $$
  \mathbf{A}^\top\mathbf{A},\mathbf{v}_i=\sigma_i^2,\mathbf{v}_i,
  $$
  where $\mathbf{v}_i$ is column $i$ of $\mathbf{V}$.

---

## 2) Relationship with $\mathbf{A}\mathbf{A}^\top$

Now form $\mathbf{A}\mathbf{A}^\top$:

$$
\mathbf{A}\mathbf{A}^\top
=========================

# (\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top)(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top)^\top

(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^\top)(\mathbf{V}\mathbf{\Sigma}^\top\mathbf{U}^\top).
$$

Use orthogonality: $\mathbf{V}^\top\mathbf{V}=\mathbf{I}_n$:

$$
\mathbf{A}\mathbf{A}^\top
=========================

# \mathbf{U}\mathbf{\Sigma}(\mathbf{V}^\top\mathbf{V})\mathbf{\Sigma}^\top\mathbf{U}^\top

\mathbf{U}(\mathbf{\Sigma}\mathbf{\Sigma}^\top)\mathbf{U}^\top.
$$

Now analyze $\mathbf{\Sigma}\mathbf{\Sigma}^\top\in\mathbb{R}^{m\times m}$: it is diagonal with entries $\sigma_1^2,\dots,\sigma_p^2$ and, if $m>p$, additional zeros:

$$
\mathbf{\Sigma}\mathbf{\Sigma}^\top
===================================

\operatorname{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)\in\mathbb{R}^{m\times m}.
$$

Therefore,
$$
\mathbf{A}\mathbf{A}^\top
=========================

\mathbf{U},\operatorname{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0),\mathbf{U}^\top.
$$

**Conclusion (left side):**

* The eigenvalues of $\mathbf{A}\mathbf{A}^\top$ are the same nonzero values $\sigma_i^2$ (again padded with zeros to reach $m$ eigenvalues total).
* The eigenvectors are the columns of $\mathbf{U}$.
* For each $i\le p$,
  $$
  \mathbf{A}\mathbf{A}^\top,\mathbf{u}_i=\sigma_i^2,\mathbf{u}_i,
  $$
  where $\mathbf{u}_i$ is column $i$ of $\mathbf{U}$.

---

## 3) Summary statement (the punchline)

* The **nonzero eigenvalues** of $\mathbf{A}^\top\mathbf{A}$ and $\mathbf{A}\mathbf{A}^\top$ are identical.
* They equal the **squares of the singular values**:
  $$
  \lambda_i(\mathbf{A}^\top\mathbf{A})=\lambda_i(\mathbf{A}\mathbf{A}^\top)=\sigma_i^2,\qquad i=1,\dots,p,
  $$
  with remaining eigenvalues (if any) equal to $0$.
* The associated eigenvectors are $\mathbf{V}$ for $\mathbf{A}^\top\mathbf{A}$ and $\mathbf{U}$ for $\mathbf{A}\mathbf{A}^\top$.

If you want, next we can do the “vector-level” derivation that starts from $\mathbf{A}\mathbf{v}_i=\sigma_i\mathbf{u}_i$ and immediately yields $\mathbf{A}^\top\mathbf{A}\mathbf{v}_i=\sigma_i^2\mathbf{v}_i$ and $\mathbf{A}\mathbf{A}^\top\mathbf{u}_i=\sigma_i^2\mathbf{u}_i$ — it’s a nice bridge to intuition.
