# Derivation: Relationship between the Singular Values and the Eigenvalues
One obvious question is how are the singular values related to the eigenvalues of a matrix? Assume $\mathbf{A}\in\mathbb{R}^{m\times n}$ has the singular value decomposition (SVD):
$$
\begin{align*}
\mathbf{A} &=\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top}\\
\end{align*}
$$
with $\mathbf{U}\in\mathbb{R}^{m\times m}$ orthogonal, $\mathbf{V}\in\mathbb{R}^{n\times n}$ orthogonal, and $\mathbf{\Sigma}\in\mathbb{R}^{m\times n}$ is diagonal (nonnegative entries on its main diagonal). Let $p=\min(m,n)$ and denote the singular values as $\sigma_1,\dots,\sigma_p$.

The singular values, $\sigma_i$, are related to the eigenvalues, $\lambda_i$, of the matrices $\mathbf{A}^\top \mathbf{A}$ and $\mathbf{A}\mathbf{A}^\top$. Further, the columns of $\mathbf{V}$ (right singular vectors) are the eigenvectors of $\mathbf{A}^\top \mathbf{A}$ and the columns of $\mathbf{U}$ (left singular vectors) are the eigenvectors of $\mathbf{A}\mathbf{A}^\top$. Where does this come from?

___

## Right singular vectors and eigenvalues of $\mathbf{A}^\top \mathbf{A}$
Start by forming the (square) matrix $\mathbf{A}^\top\mathbf{A}\in\mathbb{R}^{n\times n}$. Note that $\mathbf{A}^\top\mathbf{A}$ is always symmetric:
$$
(\mathbf{A}^\top\mathbf{A})^\top = \mathbf{A}^\top(\mathbf{A}^\top)^\top = \mathbf{A}^\top\mathbf{A}
$$
and positive semidefinite: for any $\mathbf{x}\in\mathbb{R}^n$,
$$
\mathbf{x}^\top(\mathbf{A}^\top\mathbf{A})\mathbf{x} = (\mathbf{A}\mathbf{x})^\top(\mathbf{A}\mathbf{x}) = \|\mathbf{A}\mathbf{x}\|^2 \geq 0
$$
Therefore, $\mathbf{A}^\top\mathbf{A}$ has an eigendecomposition with real, non-negative eigenvalues. Let $\mathbf{Q}\in\mathbb{R}^{n\times n}$ be an orthogonal matrix of eigenvectors and $\mathbf{\Lambda}\in\mathbb{R}^{n\times n}$ be a diagonal matrix of eigenvalues such that:
$$
\begin{align*}
\mathbf{A}^\top\mathbf{A} &= \mathbf{Q}\mathbf{\Lambda}\mathbf{Q}^{\top}\\
\end{align*}
$$
We want to show that $\mathbf{Q} = \mathbf{V}$ (i.e., the eigenvectors of $\mathbf{A}^\top\mathbf{A}$ are the right singular vectors) and relate the eigenvalues in $\mathbf{\Lambda}$ to the singular values in $\mathbf{\Sigma}$.

Now let's compute $\mathbf{A}^\top\mathbf{A}$ using the SVD decomposition of $\mathbf{A}$:
$$
\begin{align*}
\mathbf{A}^\top\mathbf{A} &= (\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top})^\top(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top})\\
&= \left(\mathbf{V}\mathbf{\Sigma}^{\top}\mathbf{U}^{\top}\right)\left(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top}\right)\\
&= \mathbf{V}\mathbf{\Sigma}^{\top}\underbrace{\left(\mathbf{U}^\top\mathbf{U}\right)}_{\mathbf{I}_{m}}\mathbf{\Sigma}\mathbf{V}^{\top}\quad\text{(since $\mathbf{U}$ is orthogonal)}\\
&= \mathbf{V}\left(\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\right)\mathbf{V}^{\top}\\
\end{align*}
$$
Note that $\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\in\mathbb{R}^{n\times n}$ is a diagonal matrix with diagonal entries $\sigma_1^2,\dots,\sigma_p^2$ (the squares of the singular values) followed by $n-p$ zeros:
$$
\begin{align*}
\mathbf{\Sigma}^{\top}\mathbf{\Sigma} = \begin{bmatrix}
\sigma_1^2 & 0 & \cdots & 0\\
0 & \sigma_2^2 & \cdots & 0\\
\vdots & \vdots & \ddots & \vdots\\
0 & 0 & \cdots & 0
\end{bmatrix}
\end{align*}
$$
or in a more compact notation $\mathbf{\Sigma}^{\top}\mathbf{\Sigma} = \mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)$. 

Now we have two expressions for $\mathbf{A}^\top\mathbf{A}$:
$$
\begin{align*}
\mathbf{A}^\top\mathbf{A} &= \mathbf{Q}\mathbf{\Lambda}\mathbf{Q}^{\top}\quad\text{(eigendecomposition)}\\
\mathbf{A}^\top\mathbf{A} &= \mathbf{V}\left(\mathbf{\Sigma}^{\top}\mathbf{\Sigma}\right)\mathbf{V}^{\top}\quad\text{(from SVD)}
\end{align*}
$$
Since $\mathbf{A}^\top\mathbf{A}$ is symmetric and positive semidefinite, its eigendecomposition is unique (up to ordering and signs of eigenvectors). Comparing these two expressions, we conclude:
$$
\boxed{
\begin{align*}
\mathbf{Q} &= \mathbf{V}\\
\mathbf{\Lambda} &= \mathbf{\Sigma}^{\top}\mathbf{\Sigma} = \mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)\\
\end{align*}}
$$
Thus, the columns of $\mathbf{V}$ (the right singular vectors) are indeed the eigenvectors of $\mathbf{A}^\top \mathbf{A}$, and the corresponding eigenvalues are the squares of the singular values, $\lambda_i=\sigma_i^2$ for $i=1,\dots,p$, with the remaining $n-p$ eigenvalues being zero.

## Left singular vectors and eigenvalues of $\mathbf{A}\mathbf{A}^\top$
Now let's consider the (square) matrix $\mathbf{A}\mathbf{A}^\top\in\mathbb{R}^{m\times m}$. Note that $\mathbf{A}\mathbf{A}^\top$ is always symmetric:
$$
(\mathbf{A}\mathbf{A}^\top)^\top = (\mathbf{A}^\top)^\top\mathbf{A}^\top = \mathbf{A}\mathbf{A}^\top
$$
and positive semidefinite: for any $\mathbf{y}\in\mathbb{R}^m$,
$$
\mathbf{y}^\top(\mathbf{A}\mathbf{A}^\top)\mathbf{y} = (\mathbf{A}^\top\mathbf{y})^\top(\mathbf{A}^\top\mathbf{y}) = \|\mathbf{A}^\top\mathbf{y}\|^2 \geq 0
$$
Therefore, $\mathbf{A}\mathbf{A}^\top$ has an eigendecomposition with real, non-negative eigenvalues. Let $\mathbf{P}\in\mathbb{R}^{m\times m}$ be an orthogonal matrix of eigenvectors and $\mathbf{\Omega}\in\mathbb{R}^{m\times m}$ be a diagonal matrix of eigenvalues such that:
$$
\begin{align*}
\mathbf{A}\mathbf{A}^\top &= \mathbf{P}\mathbf{\Omega}\mathbf{P}^{\top}\\
\end{align*}
$$
We want to show that $\mathbf{P} = \mathbf{U}$ (i.e., the eigenvectors of $\mathbf{A}\mathbf{A}^\top$ are the left singular vectors) and relate the eigenvalues in $\mathbf{\Omega}$ to the singular values in $\mathbf{\Sigma}$.

Now let's compute $\mathbf{A}\mathbf{A}^\top$ using the SVD decomposition of $\mathbf{A}$:
$$
\begin{align*}
\mathbf{A}\mathbf{A}^\top &= (\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top})(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top})^\top\\
&= \left(\mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top}\right)\left(\mathbf{V}\mathbf{\Sigma}^{\top}\mathbf{U}^{\top}\right)\\
&= \mathbf{U}\mathbf{\Sigma}\underbrace{\left(\mathbf{V}^\top\mathbf{V}\right)}_{\mathbf{I}_{n}}\mathbf{\Sigma}^{\top}\mathbf{U}^{\top}\quad\text{(since $\mathbf{V}$ is orthogonal)}\\
&= \mathbf{U}\left(\mathbf{\Sigma}\mathbf{\Sigma}^{\top}\right)\mathbf{U}^{\top}\\
\end{align*}
$$
Note that $\mathbf{\Sigma}\mathbf{\Sigma}^{\top}\in\mathbb{R}^{m\times m}$ is a diagonal matrix with diagonal entries $\sigma_1^2,\dots,\sigma_p^2$ (the squares of the singular values) followed by $m-p$ zeros:
$$
\begin{align*}
\mathbf{\Sigma}\mathbf{\Sigma}^{\top} = \begin{bmatrix}
\sigma_1^2 & 0 & \cdots & 0\\
0 & \sigma_2^2 & \cdots & 0\\
\vdots & \vdots & \ddots & \vdots\\
0 & 0 & \cdots & 0
\end{bmatrix}
\end{align*}
$$
or in a more compact notation $\mathbf{\Sigma}\mathbf{\Sigma}^{\top} = \mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)$. 

Now we have two expressions for $\mathbf{A}\mathbf{A}^\top$:
$$
\begin{align*}
\mathbf{A}\mathbf{A}^\top &= \mathbf{P}\mathbf{\Omega}\mathbf{P}^{\top}\quad\text{(eigendecomposition)}\\
\mathbf{A}\mathbf{A}^\top &= \mathbf{U}\left(\mathbf{\Sigma}\mathbf{\Sigma}^{\top}\right)\mathbf{U}^{\top}\quad\text{(from SVD)}
\end{align*}
$$
Since $\mathbf{A}\mathbf{A}^\top$ is symmetric and positive semidefinite, its eigendecomposition is unique (up to ordering and signs of eigenvectors). Comparing these two expressions, we conclude:
$$
\boxed{
\begin{align*}
\mathbf{P} &= \mathbf{U}\\
\mathbf{\Omega} &= \mathbf{\Sigma}\mathbf{\Sigma}^{\top} = \mathrm{diag}(\sigma_1^2,\dots,\sigma_p^2,0,\dots,0)\\
\end{align*}}
$$
Thus, the columns of $\mathbf{U}$ (the left singular vectors) are indeed the eigenvectors of $\mathbf{A}\mathbf{A}^\top$, and the corresponding eigenvalues are the squares of the singular values, $\lambda_i=\sigma_i^2$ for $i=1,\dots,p$, with the remaining $m-p$ eigenvalues being zero.

## Summary: Connecting SVD and Eigendecomposition

We have established the fundamental relationship between singular values and eigenvalues:

**Key Results:**
1. The eigenvalues of $\mathbf{A}^\top\mathbf{A}$ are $\lambda_i = \sigma_i^2$ for $i=1,\dots,p$, with the remaining eigenvalues equal to zero
2. The eigenvalues of $\mathbf{A}\mathbf{A}^\top$ are $\lambda_i = \sigma_i^2$ for $i=1,\dots,p$, with the remaining eigenvalues equal to zero
3. The right singular vectors $\mathbf{V}$ are the eigenvectors of $\mathbf{A}^\top\mathbf{A}$
4. The left singular vectors $\mathbf{U}$ are the eigenvectors of $\mathbf{A}\mathbf{A}^\top$

**Important Observations:**
- Both $\mathbf{A}^\top\mathbf{A}$ and $\mathbf{A}\mathbf{A}^\top$ share the same **nonzero eigenvalues** ($\sigma_1^2,\dots,\sigma_p^2$), but they may have different dimensions ($n\times n$ versus $m\times m$)
- The singular values $\sigma_i$ are the square roots of these eigenvalues: $\sigma_i = \sqrt{\lambda_i}$
- Since eigenvalues are non-negative (PSD matrices), singular values are always real and non-negative

**Intuition:**
Why does this connection make sense? The matrix $\mathbf{A}^\top\mathbf{A}$ measures how $\mathbf{A}$ stretches vectors in the **domain space** $\mathbb{R}^n$, while $\mathbf{A}\mathbf{A}^\top$ measures how $\mathbf{A}$ stretches vectors in the **codomain space** $\mathbb{R}^m$. The singular values quantify the magnitude of these stretching factors, which appear as eigenvalues (squared) in both perspectives.

**Practical Implications:**
This connection suggests two computational approaches to find singular values:
1. **Direct SVD algorithms** (QR iteration on bidiagonal form, etc.)
2. **Eigenvalue methods**: Compute eigenvalues of $\mathbf{A}^\top\mathbf{A}$ or $\mathbf{A}\mathbf{A}^\top$ and take square roots

While method 2 seems simpler, forming $\mathbf{A}^\top\mathbf{A}$ can square the condition number of $\mathbf{A}$, making it numerically less stable. Modern SVD algorithms avoid this issue by working directly with $\mathbf{A}$.
